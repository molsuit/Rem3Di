```python
from pydantic import BaseModel, AfterValidator, PlainSerializer, WithJsonSchema
from typing import Annotated
from e3nn.o3 import Irreps
```

```python
IrrepType = Annotated[
    str,
    AfterValidator(lambda v: Irreps(v)),
    PlainSerializer(lambda v: str(v)),
    WithJsonSchema(
        {
            "type": "string",
            "description": "A string representation of irreps.",
            "examples": [
                "10x0o + 5x1e + 3x2o",
            ],
        }
    ),
]

# ActivationType = Annotated[
#     Activation # this is the enum,
#     AfterValidator() # load the function from torch,
#     Serializer() # module.__class__
# ]
```

```python
class IrrepsConfig(BaseModel):
    in_irreps: IrrepType
```

```python
InputIrreps = IrrepsConfig(in_irreps="128x0e + 128x1o")
```

```python
type(InputIrreps.in_irreps)
```

```python
import pydantic_yaml as pyaml
```

```python
pyaml.to_yaml_file("irreps_config.yaml", InputIrreps)
```

```python
loaded_config = pyaml.parse_yaml_file_as(IrrepsConfig, "irreps_config.yaml")
```
