# Get descriptors from a published model

Load a Rem3Di model directory and embed a dataset into descriptor vectors, using `RemediCalculator` (the tested one-object path).

**Needs:** a published model directory and a MoleculeDataset zarr (edit the two paths below). See [docs/quickstart.md](../docs/quickstart.md) and [docs/concepts.md](../docs/concepts.md).

```python
MODEL_DIR = "/path/to/published_model"  # <-- edit
DATASET = "/path/to/dataset_zarr"  # <-- edit
DEVICE = "cuda"  # or "cpu"
```

```python
from remedi.data_handling.dataset.molecule_dataset import MoleculeDataset
from remedi.evaluation.benchmark.descriptors import RemediCalculator

calc = RemediCalculator(model_dir=MODEL_DIR, batch_size=64, device=DEVICE)
dataset = MoleculeDataset.open_existing_dataset_from_dir(DATASET)
X = calc.calculate(dataset)  # numpy array (N_molecules, D)
print(X.shape)
```

`X` is ready for [`02_train_downstream_head.ipynb`](02_train_downstream_head.ipynb).

Moved the checkpoint between machines? The baked-in MACE path may not exist locally — pass `RemediCalculator(model_dir=..., mace_model_path="/local/MACE.model")`.

You can also embed structures you build in memory: `calc.embed_atoms([atoms_a, atoms_b], total_charge=0.0, multiplicity=1.0)`.
